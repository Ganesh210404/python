```python
def book_tickets(name,from_city,to_city,seats):
    print(f"{name}:{from_city} -> {to_city} {seats}seats")
book_tickets('Alice','Delhi','Mumbai',2)
book_tickets(name="Ganesh",from_city="Knr",to_city="Hyd",seats="2")
```
