```python
#args — lets a function accept any number of positional arguments. They all get collected into a tuple.
# **kwargs — lets a function accept any number of keyword arguments. They all get collected into a dictionary.
```

```python
#*args is like a shopping bag — you can put as many items in it as you want. **kwargs is like a labeled bag — each item has a name tag on it.

def add(*args):
    print(args)  #ITS A TUPLE
    return sum(args)
print(add(1,2,3))
```
