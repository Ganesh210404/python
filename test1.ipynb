```python
n=int(input())
c=0
for i in range(1,n+1,2):
    c+=1
    if c>1:
        print(", ",end="")
    if i%5==0:
        print("divisible by 5",end="")
    else:
        print(i,end="")
```

```python
s=int(input())
e=int(input())
a=s
b=e
if s>e:
    a,b=e,s
for i in range(a,b+1):
    if i<b:
        print(f"{i}*{i+1}",end=",")
    else:
        print(f"{i}*{i+1}")
for i in range(a,b+1):
    if i<b:
        print(i*(i+1),end=",")
    else:
        print(i*(i+1))
```
