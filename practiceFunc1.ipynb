```python
s = int(input())
e = int(input())

if s > e:
    s,e=e,s
for i in range(s,e+1):
    if i < e:
        print(f"{i}*{i+1}", end=", ")
    else:
        print(f"{i}*{i+1}")

for i in range(s,e+1):
    if i<e:
        print(i * (i + 1), end=", ")
    else:
        print(i * (i + 1))
```

```python
year=int(input("Enter a year: "))
if (year%4==0 and year%100!=0) or (year%400==0):
    print(f"{year} is a Leap Year")
else:
    print(f"{year} is Not a Leap Year")
```
